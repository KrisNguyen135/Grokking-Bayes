```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml

co2 = fetch_openml(data_id=41187, as_frame=True)
co2.frame.head()
```

```python
data = co2.frame.groupby("year")["co2"].mean().values
data
```

```python
plt.plot(data);
```

```python
np.save("./co2.npy", data)
```
